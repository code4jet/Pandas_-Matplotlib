```python
import pandas as pd
import numpy as np
obj=pd.Series([6,7,-5,3],index=['d','c','a','b'])
print(obj)

obj2=obj.reindex(['a','b','c','d','e'])
print(obj2)
```

```python
# For ordered data like time series , it may be desiable to do something interpolation or filling of values when reindexing 
obj3=pd.Series(['blue','purple','yellow'],index=[0,2,4])
print(obj3,"\n")

# "ffill" for filling with last valid  forward value and "bfill for filling with first valid backward value ." 
obj3.reindex(range(7),method='ffill')
```

```python
df=pd.DataFrame(np.arange(9).reshape((3,3)),index=['a','c','d'],columns=['one','two','three'])
print(df)
df2=df.reindex(['a','b','c','d'])
print(df2)

```

```python
# column can be reindexed with the column keyword 
s=pd.Series(['one' , 'two' , 'three','four'])
l=df2.reindex(columns=s)
print(l)
```

```python
#  Instead of NaN, fill new labels with a specific value.

k = pd.Series([10, 20, 30], index=[0, 2, 4])

k
k.reindex([0,1,2,3,4,5],fill_value=0)
```

```python
# Limits how many consecutive missing labels can be filled when using a filling method.
# limit arguement only working with "method" 
j=pd.Series(np.arange(5),index=[5,6,7,8,9])
j
j.reindex(np.arange(9),method="bfill",limit=2)
```

- `tolerance` sets the **maximum distance allowed** between requested labels and existing index labels when reindexing with `method='nearest'`, `ffill`, or `bfill`.
- If the nearest index is **within tolerance** → value is filled.
- If it’s **outside tolerance** → result is `NaN`.

```python
s = pd.Series([10, 20, 30], index=[1, 5, 10])

# Without tolerance
print(s.reindex([0, 2, 6, 9], method='nearest'))
# → Always picks the closest index

# With tolerance=0.5
print(s.reindex([0, 2, 6, 9], method='nearest', tolerance=1))
# → Too far → NaN
```
