## Dropping is use for :<br>
- Removing unnecessary columns <br> 
- Removing duplicate rows 
- Data cleanning for ML
<br> 
we use drop function for it :  `obj.drop(label , axis = 0)` 
### Parameters:
* **labels** $\rightarrow$ Row/column name(s)
* **axis = 0** $\rightarrow$ Remove rows (default)
* **axis = 1** $\rightarrow$ Remove columns
* **inplace = True** $\rightarrow$ Modify original DataFrame
* **errors = 'ignore'** $\rightarrow$ Ignore error if label doesn't exist
*note: inplace=True change original 

```python
import numpy as np
import pandas as pd
data={
    "c1":["A","B","C","D","E","F","G","H","I"],
    "c2":[55,33,22,55,22,11,44,66,77],
    "c3":["n","o","n","n","n","o","n","o","o"]
}
df=pd.DataFrame(data)
df
```

```python
# ndf=df.drop(5)
# ndf
ndf=df.drop("c1",axis=1,errors="ignore")
ndf
```
