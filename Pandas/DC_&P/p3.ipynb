### The `re` module functions fall into three categories: 
`pattern matching` , `substitution` , and `splitting` . 

```python
import re

text='jics  new\nfour\t starc'
re.split(r'\s+',text)
```

## Creating a regex tool.

```python
regex=re.compile(r'\s+')
regex.split(text)
```

```python
regex.findall(text)
```

### To Avoid escaping with \ in a regular expression , use raw string literals like r'C:\x'

Pattern
   │
   ▼
re.compile()
   │
   ▼
Regex Object
   │
   ├── findall()
   ├── search()
   ├── match()
   ├── split()
   └── sub()

```python
text = """Jeck jeck@google.com
Steve steve@gmail.com
Rob rob@gmail.com
Ryan ryan@yahoo.com
"""
pattern = r'[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,4}'

# re.IGNORECASE makes dthe regex case-insensitive
regex=re.compile(pattern,flags=re.IGNORECASE)
```

```python
regex.findall(text)
```

```python
sr=regex.search(text)
sr
```

```python
text[sr.start():sr.end()]
```

### regex.match() returns None , as it only will match if the pattern occurs at the start of the string .

```python
print(regex.match(text))
```

```python
print(regex.sub('REDACTED',text)) 
```

```python
text = "Apple costs 120 and Mango costs 80"

regex = re.compile(r'\d+')

print(regex.sub("yyy", text))
```

## in new_ptr space not required 

```python
new_pat=r'([A-Z0-9._%+-]+)@([A-Z0-9.-]+)\.([A-Z]{2,4})'
regex=re.compile(new_pat,flags=re.IGNORECASE)
```

```python
m=regex.match('freedy@bright.net')
m.groups()
```

## findall()  returns a list of tuples when the patters has groups:

```python
regex.findall(text)
```

```python
print(regex.sub(r'Username: \1, Domain : \2 Suffix : \3',text))
```
