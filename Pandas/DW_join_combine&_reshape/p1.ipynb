## Combining and Merging Datasets:


```python
df1=pd.DataFrame({'key':['b','b','a','c','a','a','b'],
                  'data1':range(7)})
df2=pd.DataFrame({"key":['a','b','d'],
                  'data2':range(3)})
df1
```

### The `join key` is the column whose `values are searched` in both DataFrames to determine `which rows should be combined`.



```python
df2
```

###  Summary
- `on=`                  → join on same column(s) in both DataFrames  
- `left_on` / `right_on` → join on different column names  
- `left_index` / `right_index` → join on indices  
- Nothing given → join on common column names  

```python
# merger function connect rows 
pd.merge(df1,df2)
```

```python
pd.merge(df1,df2,on='key')
```

- *By default `merge` does an `inner` join*
- *The keys in the result are the intersection*

```python
df3=pd.DataFrame({'lkey':['b','b','a','c','a','a','b'],
                  'data1':range(7)})
df4=pd.DataFrame({"rkey":['a','b','d'],
                  'data2':range(3)})
pd.merge(df3,df4,left_on='lkey',right_on='rkey')
```

- *The `outer` join takes the union of the keys*

```python
pd.merge(df1,df2,how='outer')
```

```python
dff1 = pd.DataFrame({'A': [1, 2]})
dff2 = pd.DataFrame({'B': ['x', 'y', 'z']})
pd.merge(dff1,dff2,how='cross')
```

```python
left = pd.DataFrame({'key1': ['foo', 'foo', 'bar'],
     'key2': ['one', 'two', 'one'],
    'lval': [1, 2, 3]})
right = pd.DataFrame({'key1': ['foo', 'foo', 'bar', 'bar'],
                     'key2': ['one', 'one', 'one', 'two'],
                     'rval': [4, 5, 6, 7]})
pd.merge(left, right, on=['key1', 'key2'], how='outer')
# outer == union 
```

```python
pd.merge(left,right,on='key1')
# Match rows where key1 is equal.

# key2_x = from left
# key2_y = from right
```

```python
pd.merge(left,right,on='key1',suffixes=('_left','_right'))
```

```python
pd.merge(left, right, on='key1', how='outer', indicator=True)

# Output adds:/
```

## Merging on Index

*if we pass `left_index=True` or `right_index=True` or `both` then index should be used as the merge key*

```python
l1=pd.DataFrame({'key':['a','b','a','a','b','c'],
                 'value':range(6)})
r1=pd.DataFrame({'grp_val':[3.5,7]} ,index=['a','b'])
l1
```

```python
r1
```

```python
pd.merge(l1,r1,how='outer',left_on='key',right_index=True)

# Reason: sort=True behavior in merge only for 'key'
```

## Merging on hierarchically indexed data 

```python
lefth=pd.DataFrame({'key1':['NY','NY','NY','TOKYO','TOKYO'],
                    'key2':[2000,2001,2002,2001,2002],
                    'data':np.arange(5.)})
righth = pd.DataFrame(np.arange(12).reshape((6, 2)),
                     index=[['TOKYO', 'TOKYO', 'NY', 'NY', 'NY', 'NY'],
                            [2001, 2000, 2000, 2000, 2001, 2002]],
                    columns=['event1', 'event2'])
lefth
```

```python
righth
```

```python
pd.merge(lefth,righth,left_on=['key1','key2'],right_index=True,sort=True)
```

```python
pd.merge(lefth,righth,left_on=['key1','key2'],right_index=True,how='outer')
```

```python
left2 = pd.DataFrame([[1., 2.], [3., 4.], [5., 6.]],
                    index=['a', 'c', 'e'],
                    columns=['Ohio', 'Nevada'])
right2 = pd.DataFrame([[7., 8.], [9., 10.], [11., 12.], [13, 14]],
                    index=['b', 'c', 'd', 'e'],
                    columns=['Missouri', 'Alabama'])
```

```python
pd.merge(left2,right2,left_index=True,right_index=True,how='outer')
```

###  Summary : `join()`
- `on=` → left DataFrame column(s) + right DataFrame index  
- `how=` → specify join type (`left`, `right`, `inner`, `outer`, `cross`)  
- `lsuffix` / `rsuffix` → handle overlapping column names  
- Nothing given → default join on indices  
- Unlike `merge()`, `join()` does **not** support `right_on`, `left_on`, `left_index`, or `right_index` parameters  

```python
left2.join(right2,how='outer')
```

```python
l1.join(r1,on='key')
```

```python
another = pd.DataFrame([[7., 8.], [9., 10.], [11., 12.], [16., 17.]],
                         index=['a', 'c', 'e', 'f'],
                         columns=['New York', 'Oregon'])
another
```

```python
left2.join([right2,another])
```

```python
left2.join([right2,another],how='outer')
```
