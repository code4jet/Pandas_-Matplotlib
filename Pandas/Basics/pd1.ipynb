## Why we save DataFrames to CSV
-  To store data permanently outside Python.
-  To share datasets with others (Excel, Google Sheets, etc. can open CSV easily).

-  To reuse later without retyping or regenerating data.

-  To export results after cleaning, filtering, or analysis.

```python


# Example DataFrame
data = {
    "Student_ID": [101, 102, 103],
    "Name": ["grey","hugh","darvin"],
    "Marks": [33,55,66]
}
df = pd.DataFrame(data)

# Save to CSV
df.to_csv("students.csv", index=False)
pd.read_csv("students.csv")                     # Quick check after saving 

```
