### INDEX OBJECTS : 
- " Index objects are immutable and thus can't be modified by user "

Some index methods and properties :

```python
# 1) append 

import numpy as np 
import pandas as pd

idx1=pd.Index([1,2,3])
idx2=pd.Index([3,4,5])

x=idx1.append(idx2)
# Creates new index.
print(x)
type(x)
```

```python
# 2) difference()   : A - B

print(idx1.difference(idx2))
```

```python
# 3) intersection() and union() :

A=pd.Index([1,2,3,4])
B=pd.Index([3,4,5])
print(A.intersection(B))
print(A.union(B))
```

```python
# 4) isin() and  delete(index)

idx=pd.Index([4,5,6,7,8])
print(idx.isin([4]))
print(idx.isin([7,8]))
print(idx.delete(2))
```

```python
# 5) drop(value)  and insert(index  , value )

print(idx.drop(5))
print(idx.insert(0,7))
```

```python
# 6) is_monotonic

p=pd.Index([10,15,20,25,30])
print(p.is_monotonic_increasing)
print(p.is_monotonic_decreasing)
```

```python
# 7) is_unique 

print(idx)
print(idx.is_unique)
idx=idx.insert(2,7)
print(idx.is_unique)
```

```python
# 8) unique()

idx3=pd.Index([10,20,20,30,40,50,50,60])
print(idx3.unique())
```
