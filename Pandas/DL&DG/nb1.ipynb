### Irregular whitespace-delimited files : 

```python
with open('examples/ex3.txt', 'w') as f:
    f.write("            A         B         C\n")
    f.write("aaa -0.264438 -1.026059 -0.619500\n")
    f.write("bbb  0.927272  0.302904 -0.032399\n")
    f.write("ccc -0.264273 -0.386314 -0.217601\n")
    f.write("ddd -0.871858 -0.348382  1.100491\n")
                                                                
```

```python
result = pd.read_table('examples/ex3.txt', sep=r'\s+')
result
```

### sep=r'\s+' (regex) handles variable whitespace. If headers < data columns by exactly one,<br> pandas auto-assumes the first column is the index.

### Always use raw strings (r'\s+') for regex patterns in pandas — avoids escape sequence warnings.

## Skipping junk/comment rows — skiprows

```python
with open('examples/ex4.csv', 'w') as f:
    f.write("# hey!\n")
    f.write("a,b,c,d,message\n")
    f.write("# just wanted to make things more difficult for you\n")
    f.write("# who reads CSV files with computers, anyway?\n")
    f.write("1,2,3,4,hello\n")
    f.write("5,6,7,8,world\n")
    f.write("9,10,11,12,foo\n")
```

```python
pd.read_csv('examples/ex4.csv', skiprows=[0, 2, 3, 6])
```

### skiprows=[list of row numbers] removes specific junk/comment rows before parsing. <br>Row numbers are 0-indexed.

##  Reading large files in pieces — nrows and chunksize

- nrows=N = preview first N rows.
- chunksize=N = process large file in N-row batches,<br> useful for memory-limited aggregation.

```python
import numpy as np

rng = np.random.default_rng(0)
big_df = pd.DataFrame(
    rng.standard_normal((10000, 4)),
    columns=['one', 'two', 'three', 'four']
)
big_df['key'] = rng.choice(list('ABCDEFGHIJKLMNOPQRSTUVWXYZ'), size=10000)
big_df.to_csv('examples/ex6.csv', index=False)
```

```python
#  nrows — just read the first N rows
r1=pd.read_csv('examples/ex6.csv', nrows=5)
r1

# Loads only the first 5 rows — fast, useful for previewing a huge file.
```

```python
# chunksize — read the file in pieces (for processing)

chunker = pd.read_csv('examples/ex6.csv', chunksize=1000)
chunker

# it returns a TextFileReader object you can iterate over.
# Each iteration gives you a DataFrame of 1000 rows at a time.
```

## get the top 10 most frequent letters in key, computed without ever loading all 10,000 rows at once.


```python

tot = pd.Series([], dtype='float64')
for piece in chunker:
    tot = tot.add(piece['key'].value_counts(), fill_value=0)

tot = tot.sort_values(ascending=False)
tot[:10]    
```

## Working with Delimited Formats 

```python
with open('examples/ex7.csv', 'w') as f:
    f.write('"a","b","c"\n')
    f.write('"1","2","3"\n')
    f.write('"4","5","6"\n')
```

```python
# For any file with a single-char delimeter , we can use python's built-IN CSV module 

import csv

f=open("examples/ex7.csv")
reader=csv.reader(f)
```

```python
for line in reader:
    print(line)

# quote characters automatically stripped — not a DataFrame, just raw rows.
```

```python
# how to use csv module perfectly

f =  open('examples/ex7.csv')
lines = list(csv.reader(f))

header, values = lines[0], lines[1:]
print(header)
print(values)
```

```python
data_dict = {h: v for h, v in zip(header, zip(*values))}
data_dict

# all column give in form of rows(dict's key :value)
```

## Custom dialects for weird delimiters/quoting

```python
class my_dialect(csv.Dialect):
    lineterminator="\n"
    delimiter=";"
    quotechar='"'
    quoting=csv.QUOTE_MINIMAL

with open('examples/ex7.csv','r') as g:
    reader = csv.reader(g, dialect=my_dialect)
    print(list(reader))
```

```python
with open('examples/ex7.csv', 'r') as g:
    reader = csv.reader(g)   # default delimiter=',' — matches the file
    print(list(reader))
```

## Writing delimited files manually — csv.writer

```python
with open('examples/mydata.csv', 'w') as f:
    writer = csv.writer(f, dialect=my_dialect)
    writer.writerow(('one', 'two', 'three'))
    writer.writerow(('1', '2', '3'))
    writer.writerow(('4', '5', '6'))
    writer.writerow(('7', '8', '9'))
```
