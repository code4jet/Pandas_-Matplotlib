# Statistical Moments

A **statistical moment** is a mathematical measure that describes a specific property of a dataset or probability distribution.

Each higher-order moment provides additional information about the distribution.

| Moment | Power | Measures | Statistical Measure | What it tells |
|:------:|:-----:|----------|---------------------|---------------|
| **1st Moment** | 1 | Center | Mean (μ) | Where the data is centered (average value). |
| **2nd Moment** | 2 | Spread | Variance (σ²) / Standard Deviation (σ) | How spread out the data is around the mean. |
| **3rd Moment** | 3 | Asymmetry | Skewness | Whether the distribution is tilted to the left or right. |
| **4th Moment** | 4 | Tail Heaviness | Kurtosis | How heavy or light the tails are (tendency to produce extreme values). |

### What is ddof?

- **ddof** stands for `*Delta Degrees of Freedom*`  
- It tells pandas what number to subtract from **N** in the denominator  
- Used in variance and standard deviation calculations  
- Formula becomes:


$$
\text{Variance} = \frac{\text{Sum of squared deviations}}{N - ddof}
$$



👉 Common case: `ddof=1` → denominator becomes \(N-1\) (sample variance)

### skew()

**Purpose:**  
Measures the asymmetry (lack of symmetry) of a distribution.  
It tells you which side has the longer tail.

- skew = 0 → Symmetric  
- skew > 0 → Right-skewed (tail on the right)  
- skew < 0 → Left-skewed (tail on the left)
- Skewness measures how asymmetric (uneven) a distribution is.

## case 1
*value=0*

- You can say:

Distribution is approximately symmetric.
Left and right sides are almost balanced.<br>
`Mean ≈ Median ≈ Mode.`

### Case 2
*2.3 > 0*

- You can say:

Data is positively (right) skewed.
There are some unusually large values (high outliers).<br>
`Mean > Median.`
Long tail on the right.

- example: salary <br>

### Case 3
*-1.7 < 0*

- You can say:

Data is negatively (left) skewed.
There are some unusually small values.
Mean < Median.
Long tail on the left.

-Example : <br>

```python
s = pd.Series([2, 4, 6, 8, 50])
print(s.skew())

# Returns => Positive value
# Meaning => Right-skewed
```

```python
s = pd.Series([1, 40, 42, 44, 46])
print(s.skew())

# Returns => Negative value
# Meaning => Left-skewed

```

### Skewness Interpretation

| Skewness Value | Interpretation |
|---|---|
| **-0.5 to +0.5** | Approximately **symmetric** |
| **-1 to -0.5** | **Moderately negatively skewed** |
| **+0.5 to +1** | **Moderately positively skewed** |
| **< -1** | **Highly negatively skewed** |
| **> +1** | **Highly positively skewed** |

### ML / EDA Usage of skew()

- Very useful before training models  
- Large skewness → feature not normally distributed  
- Transformations to reduce skewness:
  - `np.log()`  
  - `np.sqrt()`  
  - **Box-Cox**  
  - **Yeo-Johnson**


 Skewness (3rd Standardized Moment)

### Formula

 Skewness Formula

$$
\boxed{
\text{Skewness}
=
\frac{\frac{1}{n}\sum_{i=1}^{n}(x_i-\mu)^3}{\sigma^3}
}
$$

### Where

- $n$ = Number of observations
- $x_i$ = Each data value
- $\mu$ = Mean
- $\sigma$ = Standard deviation
### Interpretation

- **Skewness = 0** → Symmetric distribution
- **Skewness > 0** → Right (Positive) skew
- **Skewness < 0** → Left (Negative) skew

$$
\boxed{
\text{Kurtosis}
=
\frac{\frac{1}{n}\sum_{i=1}^{n}(x_i-\mu)^4}{\sigma^4}
}
$$

### kurt()

**Purpose:**  
Measures how heavy (or light) the tails of a distribution are compared to a normal distribution.  
kurtosis measures how heavy the tails are, which tells us how many extreme values (outliers) are present.
- **Tails** = values far away from the center (extreme values).  
- Most data usually lies near the middle, but tails show how far extreme values stretch.

- Most students are near 50, but **5, 10, 95, 100** are far away.  
- Graph stretches farther → **heavy tails**.  
- Called "heavy" because **more data lives in the tails** (more extreme values than expected).

#### What does `kurt()` tell us?
It answers: **"How many extreme values (outliers) does this data have?"**

- **High kurtosis** → more extreme values, heavy tails  
  Example: `1, 2, 3, 4, 5, 6, 100` → one huge outlier → high kurtosis  
- **Low kurtosis** → fewer extreme values, light tails  
  Example: `48, 49, 50, 50, 51, 52` → everything close → low kurtosis  
- **Normal kurtosis (~0)** → distribution similar to normal curve

---

👉 In **EDA/ML**, kurtosis helps detect **outliers** and understand tail behavior of features before training models.


```python
# Measures how heavy the tails of the distribution are.

s1=pd.Series([5, 10, 48, 49, 50, 51, 52, 95, 100])
print(s1.kurt())
s2=pd.Series([48, 49, 49, 50, 50, 50, 51, 51, 52])
print(s2.kurt())

s3=pd.Series ([ -100 ,48, 49, 50, 50, 51, 52, 300])
print(s3.kurt())
```

kurt() < 0 :
Less outliers
Light tails

kurt() = 0 :
Normal outliers
Normal tails

kurt() > 0 :
More outliers
Heavy tails

| Kurtosis Value | Interpretation                                              |
| -------------- | ----------------------------------------------------------- |
| **≈ 0**        | **Mesokurtic** — similar to normal distribution             |
| **> 0**        | **Leptokurtic** — heavier tails / more extreme values       |
| **< 0**        | **Platykurtic** — lighter tails / fewer extreme values      |

### .cummax()  
**Purpose:**  
Returns the maximum value up to each row (cumulative maximum).  
It keeps track of the largest value encountered so far, row by row.  

### .cummin()  
**Purpose:**  
Returns the minimum value up to each row (cumulative minimum).  
It keeps track of the smallest value encountered so far, row by row.  


```python
# Running maximum.
# Running minmum.

print(d2)
print(d2.cummax())
print(d2.cummin())
```

### .diff()

**Purpose:**  
Returns the difference between the current element and the previous element.  
It shows how much change happened row by row.

- Positive value → increase compared to previous row  
- Negative value → decrease compared to previous row  


```python
print(d2.diff())

#    Math  Science  English
# 0   NaN     NaN     NaN   # first row has no previous
# 1 -10.0    10.0    -6.0   # 80-90, 95-85, 82-88
# 2  10.0   -10.0     9.0   # 90-80, 85-95, 91-82
# 3 -20.0   -10.0    -3.0   # 70-90, 75-85, 88-91
# 4   NaN    15.0    -3.0   # NaN-70, 90-75, 85-88

print(d2["Math"].diff())


# .diff(period=1) → current row minus previous row

# .diff(period=2) → current row minus 2 rows before

# .diff(period=n) → current row minus n rows befor
```

```python
newsr=pd.Series([80,80,80,90,80,80,70,70])
newsr.diff()
```
