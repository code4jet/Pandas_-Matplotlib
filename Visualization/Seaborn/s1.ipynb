### 1. Relational Plot

- to see the statistical relation between 2 or more variables.
- Bivariate Analysis

Plots under this section

- scatterplot
- lineplot

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import plotly.express as px


year = [2020, 2021, 2022, 2023, 2024]
runs = [500, 650, 720, 580, 800]

fig = px.line(x=year, y=runs)

fig.show()
```

### 2. Distribution Plots

- used for univariate analysis
- used to find out the distribution
- Range of the observation
- Central Tendency
- is the data bimodal?
- Are there outliers?

Plots under distribution plot

## This is axis level functions : 

- histplot
- kdeplot
- rugplot

## Figure level Function : `displot`

```python
# Univariate analysis 
tips
```

```python
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(kind='hist',data=tips,x='total_bill')
```

```python
sns.displot(kind='hist',data=tips,x='total_bill',bins=4)
```

```python
#  sns.histplot can absolutely be used to plot a categorical column,
#  and it does so by treating the unique categories as discrete bins.


# here histplot act like countplot 
sns.displot(kind='hist',data=tips,x='day')
```

```python
sns.displot(kind='hist',data=tips,x='tip',hue='sex')
```

```python
sns.displot(kind='hist',data=tips,x='tip',hue='sex',element='step')
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

```python
sns.displot(kind='hist',data=titanic,x='age',hue='sex',element='step')
```

```python
# Faceting only allows on figure level function 
sns.displot(kind='hist',data=titanic,x='age',col='sex',element='step')
```

## kdeplot : 
### Rather than using discrete bins, a KDE plot smooths the observations<br>with a Gaussian kernel, producing a continuous density estimate

```python
sns.displot(kind='hist',data=tips,x='total_bill')
```

```python
sns.kdeplot(data=tips,x='total_bill',hue='sex')
```

```python
# Size change through fig level function 

sns.displot(kind='kde',data=tips,x='total_bill',hue='sex',fill=True ,height=10,aspect=2)
```

## Rugplot
## plot marginal distribution by drawing ticks along x and y axes.
## `shows density near on axis` 

```python
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# Bivariate histrogram

# A Bivariate histrogram bins the data within rectangles that tile the plot 
# and then shows the count of observation within each rectangle with the fill color  

sns.histplot(data=tips,x='total_bill',y='tip')
```

```python
# Bivariate kdeplot 
# a bivariate KDE plot smothes the (x,y) obeservation with a 2D Gaussian 

sns.kdeplot(data=tips,x='total_bill',y='tip') 
```

## Matrix plot 
- Heatmap
- Clustermap

```python
# Heatmap

# Plot rectangular data as a color-encoded matrix 

temp_df=gap.pivot(index='country',columns='year',values='lifeExp')
# temp_df

plt.figure(figsize=(10,8))
# axis level function 
sns.heatmap(data=temp_df)
```

```python
# annot parameter in heatmap

plt.figure(figsize=(15,15))
dd1= gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
sns.heatmap(dd1,annot=True,linewidths=0.5,cmap='viridis')


```

```python
# clustermap

iris=px.data.iris()
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```
