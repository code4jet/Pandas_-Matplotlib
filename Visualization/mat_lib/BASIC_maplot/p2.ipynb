## Pie Chart
- <h3> Univariate/ Bivariate Analysis<br>
- <h3>Categorical vs numerical <br>
- <h3> Use case - To find contribution on standard scale <br>

```python
data=[20,45,67,78,100]
l=['eng','sci','mat','com','guj']
plt.pie(data,labels=l)
plt.show()
```

```python
df = pd.read_csv('data_sets/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%1.1f%%',labeldistance=1.2)
plt.show()
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',labeldistance=1.35,pctdistance=0.8,colors=['red','green','blue','pink','orange','yellow'])
plt.show()
```

### Explode and shadow parameter 

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.1,0,0,0,0,0.1])
plt.show()
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.3,0,0,0,0,0.3],shadow=True)
plt.show()
```
