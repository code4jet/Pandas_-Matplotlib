# Types of plots :


## Scatter plots : scatter = Separate dots = See relationship between two numeric variables.

- `Bivariate Analysis`
- `numerical vs numerical`
- `Use case - Finding correlation`

```python
x=np.linspace(-10,10,50)
y=10*x + 3 + np.random.randint(0,300,50) 
plt.scatter(x,y)
plt.show()
```

```python
np.corrcoef(x,y)
```

## Bar charts :
- `Bivariate analysis`
- `Numerical vs categorical `
- `Use case - aggregate analysis of groups `


```python
children=[20,10,40,10,30]
color=['blue','green','red','yellow','black']
plt.bar(color,children,color='b')
```

## 3 graphs side by side 
plt.bar(x - w, y1, width=w)<br>
plt.bar(x,     y2, width=w)<br>
plt.bar(x + w, y3, width=w)

```python
plt.bar(np.arange(df2.shape[0])-0.2  ,df2['2015']   ,width=0.2,label='2015')
plt.bar(np.arange(df2.shape[0])     ,df2['2016']    ,width=0.2,color='b',label='2016')
plt.bar(np.arange(df2.shape[0])+0.2  ,df2['2017']   ,width=0.2,color='g',label='2017')
plt.legend()

# we can add names by replacing numbers 
plt.xticks(np.arange(df2.shape[0]),df2['batsman'])
# plt.xticks(rotation='vertical')
plt.show()
```

```python
children=[20,10,40,10,30]
color=['blue blue blue blue blue blue ','green green green greengreen green','red','yellow yellow yellow','black black black black']
plt.bar(color,children,color='b')
plt.xticks(rotation='vertical')
plt.show()
```
