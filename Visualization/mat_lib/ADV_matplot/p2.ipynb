```python
#sharex share the x axis of both graphs 


fig, ax=plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(9,7))
ax[0].scatter(batters['avg'],batters['strike_rate'],color="green")
ax[1].scatter(batters['avg'],batters['runs'])

ax[0].set_title('Avg vs SR')
ax[0].set_ylabel('Strike rate')
ax[1].set_title('Runs vs avg')
ax[1].set_ylabel('runs')
```

## 3D Surface Plot

-> Need x, y, z<br>
x and y usually start as 1D arrays<br>

- meshgrid() creates 2D coordinate grids<br>
- `Z = f(X,Y)`<br>
- X.shape == Y.shape == Z.shape<br>
- plot_surface() draws the surface
- cmap adds color based on height

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)

xx,yy=np.meshgrid(x,y)
z=xx**2+yy**2
```

```python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

# Contour plots

```python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

## Colorbar shows depth(3d axis)

```python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z=np.sin(xx) + np.cos(yy)
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

# Heatmaps

```python
delivery = pd.read_csv("data_sets/IPL_Ball_by_Ball_2008_2022.csv")
delivery.head()

```

```python
temp_df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run']==6)]
temp_df.head()
```

```python
grid= temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')

grid
```

## `plt.xticks()` and `plt.yticks()` control the tick marks and their labels on the x-axis and y-axis of a matplotlib plot.

```python
plt.figure(figsize=(12,8))
plt.imshow(grid)
plt.yticks(delivery['overs'].unique(), list(range(1,21)))
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.colorbar()
plt.xlabel('ball_no')
plt.ylabel('over no')
plt.plot()
 
```
