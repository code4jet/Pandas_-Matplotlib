# Pandas plot
## Series

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

s = pd.Series(np.arange(1,8))
s.plot(kind='line')
```

```python
s = pd.Series(np.arange(1,8))
s.plot(kind='hist')
```

```python
tips=sns.load_dataset('tips')
tips.head()
```

```python
tips['size']=tips['size']*100
tips.head()
```

```python
# pandas -> scatter -> labels -> marker ,color (c) -> cmap
tips.plot(kind='scatter',x='total_bill',y='tip',title='cost_analysis',marker='d',figsize=(10,6),s='size',c='sex',cmap='viridis')
```

## 2D plots

```python
stocks = pd.read_csv('https://raw.githubusercontent.com/m-mehdi/pandas_tutorials/main/weekly_stocks.csv')
stocks
```

```python
# lineplot
stocks['MSFT'].plot(kind='line')
```

```python
stocks.plot(kind='line',x='Date')
```

```python
stocks[['Date','MSFT','FB']].plot(kind='line',x='Date')
```

```python
stocks[['Date','MSFT','AAPL']].plot(kind='line',x='Date')
```

```python
tips.groupby('sex')['total_bill'].mean().plot(kind='bar')
```

```python
t1['2015'].plot(kind='bar')
```

```python
# These too easy compare to matplotlib barplotting side by side for each column
t1.plot(kind='bar')
```

```python
t1.plot(kind='bar',stacked=True)
```

```python
stocks.plot(kind='hist',bins=40)
```

```python
# pie -> single and multiple
df = pd.DataFrame(
    {
        'batsman':['Dhawan','Rohit','Kohli','SKY','Pandya','Pant'],
        'match1':[120,90,35,45,12,10],
        'match2':[0,1,123,130,34,45],
        'match3':[50,24,145,45,10,90]
    }
)

df.head()
```

```python
df['match1'].plot(kind='pie',labels=df['batsman'],autopct='%0.1f%%',explode=[0,0.4,0,0.7,0,0])
```

```python
df[['match1','match2','match3']].plot(kind='pie',subplots=True,figsize=(15,8))
```

```python
# Multiple Graphs together 
stocks.plot(kind='line',subplots=True)
```

```python
tips.pivot_table(index=['day','time'],columns=['sex','smoker'],values='total_bill',aggfunc='mean').plot(kind='line')
```
