## Permutation and Random Sampling 

`A permutation is an arrangement of objects where the order matters.`

```python
df=pd.DataFrame(np.arange(5 * 4).reshape((5,4)))
df

sampler=np.random.permutation(5)
sampler
```

```python
#  take function is equivalent iloc based indexing  
df.take(sampler)
```

```python
df.sample(n=3)
```

`np.random.permutation()` → Shuffle all elements exactly once.<br>
`Series.sample(replace=True)` → Randomly pick elements; duplicates are allowed.<br>
`Series.sample(replace=False)` → Randomly pick without duplicates.

```python
# The logic of this code is: indices are randomly sampled from 0 (to len(choice) - 1) with replacement,
#  and the corresponding values are selected from choice. The resulting Series has a length of 8.
choice=pd.Series([5,7,-1,6,4])

draws=choice.sample(n=10,replace=True)
draws
```

## Computing Indicator / Dummy Variables

```python
df=pd.DataFrame({ 
    'name':['kiran','lornange','Halland','yamal','Jics','Hileo','catq'],
    'class':['c1','c3','c1','c2','c2','c1','c3']

       })
df
```

```python
print(pd.get_dummies(df['class']))
print(pd.get_dummies(df['class']   , dtype=int) )
```

```python
dum=pd.get_dummies(df['class'],prefix='class')
df_with_dum=df[['name']].join(dum)
df_with_dum
```

```python
mn=['movie_id','title','genres']
movies=pd.read_table('datasets/movies.dat',sep='::',header=None,names=mn,engine='python',
                     encoding='latin1')
movies.head(10)
```

```python
all_genres=[]
for x in movies.genres:
    all_genres.extend(x.split('|'))

genres=pd.Series(all_genres).unique()
genres
```

* For construct the indicator DF : first step--> DF with all zeros 

```python

zero_matrix = np.zeros((len(movies), len(genres)))

dummies = pd.DataFrame(zero_matrix, columns=genres)
```

```python
gen=movies.genres[0]
gen.split('|')
```

```python
# It will be find the index of gen.split() in dummies . 
dummies.columns.get_indexer(gen.split('|'))
```

```python
for i, gen in enumerate(movies.genres):
    indices=dummies.columns.get_indexer(gen.split('|'))
    dummies.iloc[i,indices]=1
    # set the value=1 through iloc  
    
```

```python
movies_windic=movies.join(dummies.add_prefix('Genre__'))
movies_windic.iloc[0]
```

### A useful techinque for statistical applications is to combine `get_dummies` with a discretization function like ` cut`:

```python
np.random.seed(42)
values=np.random.rand(10)
values
```

```python
bins=[0,0.2,0.4,0.6,0.8,1]
pd.get_dummies(pd.cut(values,bins),dtype=int)
```
