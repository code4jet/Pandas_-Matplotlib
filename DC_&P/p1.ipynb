# HANDLING MISSING VALUES 


```python
import pandas as pd
from numpy import nan as NA

sd=pd.Series(["surat","bardoli","GN",NA,"A'bad",])
sd.isnull() 
```

```python
sd[1]=None

print(sd.isnull())
print("total nulls :",sd.isnull().sum())
```

```python
print(sd.dropna())
print(sd[sd.notnull()])
```

```python
import numpy as np
df3=pd.DataFrame(np.random.randn(7,3))
df3
```

```python
df3.iloc[:4,1]=NA
df3.iloc[:2,2]=NA
df3
```

```python
# Keep a row (or column) only if it has at least thresh non-missing (non-NaN) values.
df3.dropna(thresh=2)
```

```python
# fill mssing value with 0.
df3.fillna(0)
```

```python
# fill different value in columns 
df3.fillna({1:2,2:0})
```

```python
df4=pd.DataFrame(np.random.randn(6,3))
df4.iloc[2:,1]=NA
df4.iloc[4:,2]=NA
df4
```

### Old way (Pandas < 3.0)
- `df.fillna(method="ffill")`
- `df.fillna(method="bfill")`

### New way (Pandas 3.0+)
- `df.ffill()`
- `df.bfill()`

```python
# forward fill 
df4.ffill()
```

```python
# forward fill with limit.
df4.ffill(limit=2)
```

```python
df
```

### keep arguement 


```python
df = pd.DataFrame({
    "Name": ["Raj", "Amit", "Raj", "Neha", "Raj"],
    "Age":  [20, 21, 20, 22, 20],
    "c3" :['a','a','b','c','d']
})

df
```

## TRANSFORMING DATA USING FUNCTION OR MAPPING 

```python
data=pd.DataFrame({'Players':['VK','KW','MSD','ABD','MG','JR','FD'],
                   'cars':[5,3,7,2,8,2,6]
                   })
data
```

```python
player_to_nation={
    "vk":'ind',
    "msd":'ind',
    'jr':'eng',
    'abd':'rsa',
    'kw':'nz',
    'fd':'rsa',
    'mg':'nz'
}
```

```python
l_c=data['Players'].str.lower()
l_c
```

```python
data['nation']=l_c.map(player_to_nation)
data
```

```python
data['Players'].map(lambda x:player_to_nation[x.lower()])
# data
```

# Discretizatin and Binning 

```python
ages=[20,22,25,27,21,23,37,31,61,45,41,32]
bins=[18,25,35,60,100]
cats=pd.cut(ages,bins)

cats
```

```python
cats.codes
```

```python
cats.categories
```

```python
cats.value_counts()
```

```python
type(cats)
```

```python
#  we can diretly use bins and chnage the side of interval notation :
pd.cut(ages,[18,26,36,61,100],right=False)
```

## working of `.cut()`:
Find minimum
↓
Find maximum
↓
Compute range
↓
Divide range equally
↓
Assign each value to interval

```python
grp_names=['Youth','YouthAdault','MiddleAged','Senior']
pd.cut(ages,bins,labels=grp_names)
```

```python
data=np.random.rand(20)
print(data)

pd.cut(data,4,precision=2)
# precision for digits  
```

```python
dn=np.random.randint(5,25,10)
dn
```

```python
print(pd.cut(dn,3,precision=1))
```

## working of `.qcut()`:
Sort values
↓
Find quantiles
↓
Create boundaries
↓
Each bin gets approximately equal observations

```python
sr=pd.Series([3,100,1,2])

print(pd.cut(sr,2))
print(pd.qcut(sr,2))
```

```python
nsr=pd.Series([5,10,150,200,1000,2000])
print(pd.cut(nsr,3))
print(pd.qcut(nsr,3))
```

```python
#  we can pass our quantilers 
print(pd.qcut(nsr,[0,0.5,1]))
```
