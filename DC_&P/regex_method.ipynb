## Regex: ([a-z]+) = letters, (\d*) = digits.

```python

print(s)
s.str.extract(r'([a-z]+)(\d*)')
```

```python
print(s.str.endswith("d"))
print(s.str.endswith("p"))
```

```python
s.str.findall(r"[a-z]")
```

```python
s.str.get(0)
```

```python
# True if string is alphanumeric only (no spaces).
s.str.isalnum()
```

```python
# Only letters .
s.str.isalpha()
```

```python
print(pd.Series(["123", "0.5", "٣"]).str.isdecimal())
print(pd.Series(["123", "0.5", "٣"]).str.isdigit())
print(pd.Series(["123", "⅕", "٣"]).str.isnumeric())
```
