## Univariate Analysis

## Age 
Conclusions : 

- Density is high in between range of  40-60.
- The age distribution is approximately symmetric and flatter, with lighter tails than a normal distribution.
- No outliers were detected.

```python
df['age'].describe()
```

```python
df.age.plot(kind='hist')
plt.xlabel('Age')
plt.show()
```

```python
df['age'].plot(kind='box')
```

```python
df['age'].plot(kind='kde')
```

```python
print(df['age'].skew())
print(df['age'].kurt())
```

```python
df[(df['age']>40) & (df['age']<60)].shape[0] / df.shape[0]
```
