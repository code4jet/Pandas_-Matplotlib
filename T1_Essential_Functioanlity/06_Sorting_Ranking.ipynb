## Ranking in Pandas

- Ranking assigns a `position (rank) to each value.`

- Unlike sorting, ranking **does not rearrange the original data**.
  It only tells where each value would appear in sorted order.

Think of an example:

Sorting  → Arrange students by marks.

Ranking → Keep students in the same order but assign their ranks.

```python
obj3=pd.Series([7,-6,5,4,2,-3,4])
obj3.rank()
```

##  Observe the Original Series

```python
# Let's sort the Series to verify how rank() is calculated.

obj4=obj3.sort_values()
print(obj4)
print(obj4.rank())
```

## Since two values are equal (4 and 4),
   ## Pandas gives both the average rank.

<h3> Average = (4 + 5) / 2 = 4.5

- It first determines the position of every value
- in sorted order and then writes those ranks back
- to the original Series.

Sorting changes the order.

Ranking does not.

`sort_values()`   :  Actually rearranges the values.

`rank() `         :  Keeps the original order unchanged.

```python
import pandas as pd

odf = pd.DataFrame({
    "Math":    [90, 80, 90, 70],
    "Science": [85, 95, 85, 75],
    "English": [88, 82, 91, 88]
})

odf
```

```python
odf.rank()
```

```python
ndf=odf.sort_values(by='Math')
print(ndf)


```

# Ranking Methods

method="average" (default)

method="min"

method="max"

method="first"

method="dense"

```python
# Method = "min"
# Every tied (same) value gets the smallest rank.


print(odf.rank())
odf.rank(method="min")

# that's why 3.5 and 3.5 becomes 3 and 3

```

```python
# Method = "max"
# Every tied (same) value gets the largest rank.


print(odf.rank())
odf.rank(method="max")

# that's why 3.5 and 3.5 becomes 4 and 4
```

```python
# Method = "first"
# If values are equal, assign ranks based on which one appears first in the original DataFrame.
print(odf)
odf.rank(method="first")

```

```python
# Method = "dense"
# Give tied values the same rank without leaving gaps.
# Like "min", but don't skip numbers.

print(odf.rank(method="min"))
odf.rank(method="dense")

```
