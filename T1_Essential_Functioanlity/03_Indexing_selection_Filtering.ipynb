### Syntax : for series -->  `obj[ <row_index> or <column label> ]`
### Syntax : for DataFrame -->

- `obj["column_label"] `     # Select column
- `obj[start:end] `          # Slice rows (Python slicing)


```python
import pandas as pd

marks = pd.Series(
    [88, 76, 91, 67, 82, 95, 73, 84, 90, 79, 86, 92, 70, 81],
    index=[
        "Rahul", "Priya", "Amit", "Sneha", "Karan", "Neha", "Arjun",
        "Pooja", "Riya", "Vikas", "Anjali", "Yash", "Krishna", "Mehul"
    ],
    name="Marks"
)

print(marks)
print(marks.values)
print(marks["Amit"])
print(marks["Yash"])
print(marks[2:8])
print(marks.values[3:9])


```

### Slicing 

- `.loc[row_labels, column_labels]`
- `.iloc[row_positions, column_positions]`

- `obj.loc[r1:r2 , c1:c2]`   # by labels
- `obj.iloc[r1:r2 , c1:c2] ` # by positions

- .loc (label-based) → slicing is inclusive of the end label.
👉 End label (r2 or c2) is included.


- .iloc (position-based) → behaves like normal Python slicing.
👉 End index (r2 or c2) is excluded.


```python
print(df.loc[1,"Branch"])     # Label base indexing 
print(df.iloc[1,2])           # Position  base indexing 

# select one column
print("Student id:\n")

print(df["StudentID"])
print(df.loc[:,"City"])        # Coliumn label needed 
print(df.iloc[:,6] ,"\n")      # Column position needed 
```

```python
# select one row :
print(df.loc[3])

print(df.iloc[3])
```

```python
# slice one row
print(df.loc[5,"Name":"CGPA"])

```

```python
# Slice column

print("Student id from 2 to 6:")
print(df.loc[2:6,"StudentID"]     , "\n")
print("Student Names from 3-8: ")
print(df.iloc[3:9,1])
# in iloc end is not included [x:end]

df.loc[:,"Name":"CGPA"]

```

```python
# Slice Sub DF 
df.loc[ 2: 7,"Name" : "Attendance" ]
# df.iloc[3:9,2:6]
```

### Boolean Indexing (or Boolean Selection).

```python
# df.Age > 18 gives boolean output

df[df.Age > 18]  #   or  df[df["Age"] > 18] 

df[df["Placement"]=="Yes"]
```

```python
#  with multiple condition
df[(df.Branch =='CE')]
df[(df.CGPA >= 8) & (df.Internship =="Yes")] 
```

```python
print(df["Age"] < 20)
print(df[df["Age"] < 20])
```

```python
print(df.loc[2,["CGPA","Branch","Age"]]     ,"\n")
print(df.iloc[2,[-1,5,6]]                   ,"\n")
print(df.loc[[5,6,7,8] , ["Age","StudentID"]],  "\n")
print(df.iloc[[2,5,3],[3,2,1]])
```

```python
df.loc[  : ,  : "City"]
df.iloc[  : ,  : 4]
```

### `obj.at[ "row_label" , "column label" ]`
- "Label"+"Label" ==> one value 
- Returns one cell using label.



```python
print(df.at[6,"Name"])
print(df.at[3,"Attendance"])
```

### `obj.iat[ <row_position> , <column_position> ]`
- "position"+"position" ==> one value 
- Returns one cell using position.



```python
print(df.iat[0,6])
print(df.iat[5,2])
```

### `reindex()`

- Make copy unless assigned 

```python
df.reindex([3,7,9,12],columns=["Name","Branch","Semester","CGPA"])
```
