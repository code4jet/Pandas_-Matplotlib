```python
#  Arithmetic & Data Alignment
s1 = pd.Series([7.3,-2.5,3.4,1.5], index=['a','c','d','e'])
s2 = pd.Series([-2.1,3.6,-1.5,4,3.1], index=['a','c','e','f','g'])
print(s1 + s2)

# union of row and column labels; mismatches → NaN 
# Same with DF
```

```python
df1 = pd.DataFrame({
    "A":[1,2],
    "B":[3,4]
})

df2 = pd.DataFrame({
    "A":[10,20],
    "C":[30,40]
})

print(df1+df2)
```

```python
# use fill_value
sum=df1.add(df2,fill_value=0)
sum

#  if we pass fill_value=0 then it will be add all columns
#  with their value if same row column are exist then add .

# Important Note 
# fill_value=0 does not modify df1 or df2.
# It is used only during that arithmetic operation.
# Original DataFrames remain unchanged.
```

```python
print(df1 - df2     ,"\n")
df1.sub(df2,fill_value=0)
```

```python
df1.mul(df2,fill_value=0)        # df1 *df2

# here fill value = 0 coz 0 is used inplace of NaN .

```

### What is radd, rsub, rdiv?

- The r stands for "reverse" (or reflected).

```python
print(df1.radd(5))    #  5  + df1 

print(df2.rsub(2))    #  2  - df2

print(df2.div(5))     #  df1 / 5

print(df2.rdiv(5))    #  5   / df2

# other methods : 
# floordiv, rfloordiv  -->  //
# mul, rmul --> *
# pow, rpow  --> **
```

```python
# Reindex with fill_value
df1.reindex(columns=["A","B","C","D"], fill_value=0)

# Here, newly created columns (C and D) are filled with 0 instead of NaN.
```

```python
# DataFrame ↔ Series Arithmetic (Broadcasting)

df = pd.DataFrame({
    "A": [1, 2],
    "B": [3, 4]
})

s = pd.Series({"A": 10, "B": 20})

df + s

# This is called broadcasting because the one-dimensional Series is
# automatically expanded across the DataFrame without you having to write a loop.
# To broadcast across columns (match on rows) instead, use axis='index':

```
