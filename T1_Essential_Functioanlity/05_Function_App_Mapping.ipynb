```python
# Numpy unfuncs also operate with pandas objects 
import numpy as np  
import pandas as pd

fr=pd.DataFrame(np.random.randn(3,4),columns=list("pqrs"),index=list('abc'))
fr


```

```python
fr=np.abs(fr)
fr
```

```python
f = lambda x : x*x
print(fr.apply(f))
```

```python
g = lambda x: x.max()-x.min()
print(fr.apply(g)       ,"\n\n")
print(fr.apply(g,axis="columns"))   
```


### The function passed to apply() need not return a `scalar value`;<br>it can also return `a Series with multiple values`.

```python
def fun(x):
    return pd.Series([x.max(), x.min()],index=["max","min"])

fr.apply(fun)
```

```python
fmt  = lambda x : '%.2f' % x          # Return the 2 digit of each element 
fr.map(fmt)
```

```python
fr['r'].map(fmt)
```
