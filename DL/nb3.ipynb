## Binary Data Formats

'''Python DataFrame
        │
        ▼
   Binary Format
 (pickle/parquet)
        │
 Save to Disk
        │
        ▼
Binary File
        │
Later...
        │
        ▼
Read Binary File
        │
        ▼
Same DataFrame'''


### Pickle is Python's built-in module that converts any Python object into binary data so it can be stored on disk and later restored exactly as it was.

'''Python Object
      │
 Serialize
      ▼
Binary File (.pkl)
      │
 Deserialize
      ▼
Same Python Object'''

- Serialization → Object ➜ Binary file
- Deserialization → Binary file ➜ Object

```python
import pandas as pd
import numpy as np
import pickle
frame=pd.read_csv("examples/ex1.csv")
frame
```

```python
frame.to_pickle('examples/frame_pickle')
```

```python
pd.read_pickle('examples/frame_pickle')
```

```python
df = pd.DataFrame({
    "Name":["Deep","Rahul","Aman"],
    "CGPA":[8.47,8.45,9.10]
})

with open("examples/students.pkl","wb") as f:
    pickle.dump(df,f)
```

```python
with open("examples/students.pkl","rb") as f:
    df2 = pickle.load(f)

df2
```

## Using HDF5 Format

```python

frame = pd.DataFrame({'a': np.random.standard_normal(100)})
store = pd.HDFStore('mydata.h5')
```

```python
store
```

```python
store['obj1'] = frame
store['obj1_col'] = frame['a']
store['obj1']
```

```python
store.put("obj2", frame, format="table")
store.select('obj2',where=['index>=10 and index<=15'])
```

```python
store.close()
```

```python
frame.to_hdf('mydata.h5',key='obj',format='table')
```

```python
pd.read_hdf('mydata.h5',key='obj',where=['index < 5'])
```

```python
import requests
url='https://api.github.com/repos/pandas-dev/pandas/issues'
resp=requests.get(url)
resp
```

```python
data=resp.json()
data[0]['title']
```

```python
issues=pd.DataFrame(data,columns=['number','title','labels','state'])
issues
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

data = response.json()

print(type(data))

```

```python
address_df = pd.json_normalize(info["address"])

address_df.head()
```

## Interacing with Databases

```python
# Create database

import sqlite3

query="""
CREATE TABLE test
(
 a VARCHAR(20),
 b VARCHAR(20),
 c REAL,
 d INTEGER
);"""

con=sqlite3.connect('mydata.sqlite')
con.execute(query)
```

```python
data = [('jecob','US',1.25,6),
        ('yosho','korea',3.2,4),
        ('steve','sydney',1.4,5)]

stmt="INSERT INTO test VALUES(?,?,?,?)"
con.executemany(stmt,data)
```

```python
con.commit()
```

```python
cursor=con.execute('select * from test')
rows=cursor.fetchall()
rows
```

```python
cursor.description
```

```python
con.close()
```

```python
import pandas as pd
ddf=pd.DataFrame(rows,columns=[x[0] for x in cursor.description])
ddf
```

## Pandas provides one function that does all of this internally.

```python
con = sqlite3.connect("mydata.sqlite")

df = pd.read_sql_query("SELECT * FROM test", con)

df
```
