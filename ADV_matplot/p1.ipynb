```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt

iris=pd.read_csv("data_sets/iris.csv")
iris.head()
```

```python
plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.title("iris plant")
plt.plot()
```

### Sepal Length and Petal Length have a strong positive linear relationship (r ≈ 0.87)

```python
(iris["SepalLengthCm"]).corr(iris["PetalLengthCm"])
```

```python
iris.Species.unique()
```

```python
iris['Species']=iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
plt.plot()
```

## More choices are avaliable on `matplotlib cmap` 

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='plasma')
plt.plot()
```

## alpha parameter for `opacity` and colorbar to `scale`

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='plasma',alpha=0.3)
plt.colorbar()
plt.plot()
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='plasma',alpha=0.9)
plt.colorbar()
plt.plot()
```

```python
plt.figure(figsize=(16,8))
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='plasma',alpha=0.7)
plt.colorbar()
plt.plot()
```

## Annonations 

```python
batters=pd.read_csv('data_sets/batter.csv')
batters.shape
```

```python

# random_state working like np.random.seed()
sample_df=batters.head(100).sample(25,random_state=5)
sample_df
```

## EXAMPLE 1


```python
x =[1,2,3,4]
y =[5,6,7,8]
plt.scatter(x,y)
plt.text(1,5,'point1')
plt.text(2,6,'point2')
plt.text(3,7,'point3')
plt.text(4,8,'point4',fontdict={'size':12,'color':"brown"})
plt.show()
```

```python
plt.figure(figsize=(16,9))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

## Horizonatl and vertical lines for separation

```python
plt.figure(figsize=(16,9))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

# horizonal line
plt.axhline(130,color='red')
plt.axhline(140,color='green')
plt.axvline(30,color='red')

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```
