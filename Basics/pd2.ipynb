```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Name": ["Raj", "Amit", "Neha", "Riya", "Karan", "Priya", "Rahul", "Sneha"],
    "Age": [20, 21, 22, 23, 24, 20, 25, 22],
    "City": ["Surat", "Rajkot", "Ahmedabad", "Vadodara", "Surat", "Rajkot", "Bhavnagar", "Ahmedabad"],
    "Marks": [88, 76, 91, 67, 82, 95, 73, 89],
    "Branch": ["IT", "CE", "IT", "ME", "CE", "IT", "EC", "ME"]
})

df
```

### If we specify a sequence of columns , the DF columns will be arranged in that order. it's because of the **DataFrame's internal indexing and selection mechanism**, not because DataFrames are mutable.

```python
df=pd.DataFrame(df,columns=['Name','Branch','City','Marks','Age'])
df
```

### If we pass a column that isn't contained in <br>the dict, it will appear with missing values `(NaN)`  in the result.


```python
df2=pd.DataFrame(df,columns=['Name','Branch','City','Marks','Age','debt'])
df2
```

### A Column in DF can be retrieved as a Series either by dict-like `notation` or by `attribute`: 

```python
df["Name"]
# df.Name

# df['City']
df.City
```

###  Rows/Columns can be retrieved either by <br>label using `.loc` or by<br>integer position using    `.iloc.`
- .loc[x:y] → `End is INCLUDED`
-  .iloc[x:y] → `End is EXCLUDED (like normal Python slicing) `

```python
#  label base indexing  :  
#  df.loc[<row_label> , "column_label"]  --> for single element 
#  If 2nd argument was not given then loc and iloc both give whole row other wise single element 

print(df.loc[2]  ,   "\n")
print(df.loc[2,"City"]  ,"\n")

# Position base indexing : 
# df.iloc[<row_label> , <column_number>] 
print(df.iloc[2]  ,   "\n")  


print(df.iloc[2,1])
```

```python
# Columns can be modified by assignment 
df['debt']=56
# df['debt']=np.arange(8)
df
```

```python
val=pd.Series([4,-6,8,9],index=[1,3,5,6])
df['debt']=val
df
```

```python
#  Added new column with boolean values where City is Ahemedabad

df["new_col"]=df.City == "Ahmedabad"
df
```

```python
# using del key word we can delete
del df["new_col"]    # 'new_col'  column gone permanently.
print(df.columns)
df
```

```python
#  using drop function we can delete whole row form DF
df_copy=df.drop(2)   
print(df_copy.index)
df
```

### Case 1: Nested dicts with `same keys`

```python
data = {
    "Student1": {"Name": "Alex", "Age": 21, "City": "Delhi"},
    "Student2": {"Name": "Bob", "Age": 22, "City": "Mumbai"},
    "Student3": {"Name": "Cathy", "Age": 23, "City": "Pune"}
}

df3 = pd.DataFrame(data)
df3
```

### Case 2: Nested dicts with `different keys`

```python
data = {
    "Student1": {"Name": "Alex", "Age": 21},
    "Student2": {"Name": "Bob", "Marks": 85},
    "Student3": {"City": "Pune", "Age": 23}
}

df4 = pd.DataFrame(data).T
df4
```

### Case 3: Deep nesting `(dict inside dict inside dict)`

```python
data = {
    "Student1": {"Info": {"Name": "Alex", "Age": 21}, "Marks": {"Math": 85, "Sci": 90}},
    "Student2": {"Info": {"Name": "Bob", "Age": 22}, "Marks": {"Math": 80, "Sci": 88}}
}

df = pd.json_normalize(data)
df
```

```python


data = {
    "Name": ["Aman","Riya","Kunal","Neha","Sahil","Priya"],
    "Age": [17,19,18,22,18,20],
    "Marks": [75,85,92,60,80,92],
    "City": ["Ahmedabad","Surat","Rajkot","Surat","Baroda","Ahmedabad"]
}
df5 = pd.DataFrame(data)
df5
```

```python
# 1.)  Boolean Indexing  (masking)
df5[df5.Age > 18]
df5[df5["Age"] > 18]
df5[df5.Marks > 90]

# With multiple condition : 
df5[(df5.Age == 18) & (df5.Marks > 90)]
df5[(df5["Age"] > 18) | (df5["Marks"] > 90)]         
df5[~(df5["Marks"] > 90)]
```

```python
# 2)  isin() function 
df5[df5["City"].isin(["Ahmedabad","Surat"])]
```

```python
# 3) query() 
#  internally converts to df[df["Marks"]>80] : 
df5.query("Marks > 80")

df5.query("Age == 18 and Marks > 80")
df5.query("Age == 18 or Marks > 80")
df5.query("Age == 18 and not Marks > 80")
df5.query("not (Marks > 80)")   
df5.query("not (Age == 18 and Marks > 80)")
```

```python
# 4) sort_values()
df5.sort_values("Age")
df5.sort_values("Age",ascending=False)
# note: By defalut ascending is "True" . 
```

```python
df5.sort_values(["City","Marks"])
#  First sort City wise then sort Marks wise between same City 
df5.sort_values(["Marks","Age"])
```

```python
#  sort_index()
df5.sort_index()
# if df was previously sorted/shuffled, this puts rows back by index number.
```
