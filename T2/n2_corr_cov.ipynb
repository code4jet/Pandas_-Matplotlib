# Historical Stock Data Summary

## What is `data`?

`data` is a **Pandas DataFrame** containing **5 years of historical stock market data** for the selected companies (`AAPL`, `IBM`, `MSFT`, `GOOG`).

- **Each row** → One trading day
- **Each column** → A stock property (Open, High, Low, Close, Adj Close, Volume)
- **Index** → Trading dates
- **Columns** → MultiIndex (`Price Type` → `Ticker`)

Example:

| Date | Adj Close | Adj Close | Volume | Volume |
|------|-----------|-----------|--------|--------|
|      | AAPL | GOOG | AAPL | GOOG |
|2026-07-15|210.35|185.12|54,321,000|18,245,000|

---

## Extracted DataFrames

`price = data["Adj Close"]`

- Contains the **Adjusted Closing Price** of each stock.

`volume = data["Volume"]`

- Contains the **daily trading volume** (number of shares traded) of each stock.

`returns = price.pct_change()`

- Contains the **daily percentage return (day-to-day price change)** of each stock.

Formula:

Return = (Today's Price − Yesterday's Price) / Yesterday's Price

Example:

| Yesterday | Today | Return |
|-----------|------:|--------:|
|100|110|0.10 (10%)|
|110|99|-0.10 (-10%)|

```python

# !pip install yfinance

import yfinance as yf
import pandas as pd
import numpy as np

# Stock symbols
tickers = ["AAPL", "IBM", "MSFT", "GOOG"]

# Download all stocks at once
data = yf.download(
    tickers=tickers,
    period="5y",
    auto_adjust=False,
    progress=False
)

# Show available columns
# print(data)

# Adjusted Closing Prices
price = data["Adj Close"]

# Trading Volume
volume = data["Volume"]

# Daily Percentage Returns
returns = price.pct_change()
returns.tail()
```

```python
r2=volume.pct_change()
r2.tail()
```

```python
#corr()
# Measures how strongly two variables move together and in which direction

# syntax: s1.corr(s2)
print(returns['MSFT'].corr(returns['IBM']))
print(r2['MSFT'].corr(r2['IBM']))

```

## How to interpret Correlation

Correlation is always between **-1 and +1**.

| -1 (Opposite Direction) | 0 (No Relation) | +1 (Same Direction) |
|--------------------------|-----------------|---------------------|

Microsoft and IBM have a weak positive linear relationship in their daily returns.

## Correlation Interpretation

| Correlation Value | Interpretation                          |
|-------------------|------------------------------------------|
| 1.0               | Perfect positive                        |
| 0.8 to 1.0        | Strong positive                         |
| 0.5 to 0.8        | Moderate positive                       |
| 0.2 to 0.5        | Weak positive                           |
| -0.2 to 0.2       | Very weak / almost no linear relationship |
| -0.5 to -0.2      | Weak negative                           |
| -0.8 to -0.5      | Moderate negative                       |
| -1.0              | Perfect negative                        |


### Correlation Formula

$$
\operatorname{Corr}(X,Y) =
\frac{\operatorname{Cov}(X,Y)}{s_X \, s_Y}
$$

Where:

- \(X\) = First variable  
- \(Y\) = Second variable  
- \(\operatorname{Cov}(X,Y)\) = Covariance between \(X\) and \(Y\)  
- \(s_X\) = Standard deviation of \(X\)  
- \(s_Y\) = Standard deviation of \(Y\)  


**Idea:** Correlation is the **normalized (standardized) covariance**, giving a value between **-1** and **+1**.


```python
print(returns['MSFT'].cov(returns['IBM']))
```

```python
# DataFrame's corr() and cov() methods return a full correlation or covariance matrix as DataFrame, respectively
print(returns.corr())
print(returns.cov())

# for single shell see the row and column label for individual corr and cov
```

```python
returns.corrwith(returns.IBM)
```
